# How do STEM majors compare to the highest earning majors?

The goal is to find out how much STEM majors are earning, then compare that to the highest earning category. 


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/college_major_roi.csv")
```

```python
print(df.shape)
```

## Displaying STEM majors and sorting by ROI

```python
stem_df = df[df['major_category'] == 'STEM'].sort_values(by='roi_pct', ascending=False)

# Identifying number of unique majors

majors = sorted(stem_df['major'].dropna().unique())
print(f"Number of unique majors: {len(majors)}")
print(majors)

avg_roi_by_major = (
    stem_df.groupby('major')['roi_pct']
    .mean()
    .sort_values(ascending=False)
    .round(2)
)

# following df shows average roi percentage by major
avg_roi_by_major
```

```python
fig, ax = plt.subplots()

majors_sorted = avg_roi_by_major.index.tolist()
print(majors_sorted)

roi_sorted = avg_roi_by_major.tolist()
print(roi_sorted)

ax.barh(majors_sorted, roi_sorted,  align='center')
ax.yaxis.set_inverted(True)
ax.set_ylabel("Majors")
ax.set_xlabel("Average ROI")

plt.savefig("charts/stem_major_avg_roi.png", dpi=300, bbox_inches='tight')
plt.show()
```

- The findings show that electrical engineering is the STEM major with the highest ROI
